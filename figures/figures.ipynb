```python
from pathlib import Path

import pandas as pd

DATA_DIR_CANDIDATES = [Path("data_for_latex"), Path("../data_for_latex")]
for candidate in DATA_DIR_CANDIDATES:
    if candidate.exists():
        DATA_DIR = candidate
        break
else:
    raise FileNotFoundError("Could not find data_for_latex directory")

MEM_PROOF_UPD_FILES = {
    "bilinear_mem_blind_proof_upd": "Bilinear",
    "class_mem_blind_proof_upd": "Class group",
    "rsa_trapdoorless_mem_blind_proof_upd": "RSA trapdoorless",
    "rsa_trapdoored_mem_blind_proof_upd": "RSA trapdoored",
}

mem_proof_upd_tables = {}
for bench_id, scheme in MEM_PROOF_UPD_FILES.items():
    csv_path = DATA_DIR / f"{bench_id}.csv"
    table = pd.read_csv(csv_path).rename(
        columns={"Elements": "elements", "Time (ms)": "time_ms"}
    )
    mem_proof_upd_tables[scheme] = table[["elements", "time_ms"]].copy()

list(mem_proof_upd_tables.keys())
```
